```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
pd.set_option('display.max_rows', 100)

def sparsity_report(df, name):
    n = len(df)
    nulls = df.isna().sum()
    pct = (nulls / n * 100).round(2)
    rep = pd.DataFrame({'null_count': nulls, 'null_pct': pct, 'dtype': df.dtypes}).sort_values('null_pct', ascending=False)
    print(f"--- {name}: sparsity ---")
    print(rep)
    return rep
```

```python
osha = pd.read_csv('osha-hazards-outcome/January2015toNovember2025.csv', low_memory=False)
print('OSHA shape:', osha.shape)
osha.dtypes
```

```python
osha.describe(include='all').T
```

```python
sparsity_report(osha, 'OSHA hazards/outcomes')
```

```python
storm = pd.read_csv('storm-data/StormEvents_details-ftp_v1.0_d1950_c20260323.csv', low_memory=False)
print('Storm data shape:', storm.shape)
storm.dtypes
```

```python
storm.describe(include='all').T
```

```python
sparsity_report(storm, 'Storm events details')
```
